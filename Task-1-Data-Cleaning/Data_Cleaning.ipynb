```python
# Creating copies of dataset
df_raw = df.copy()
df_clean = df.copy()
```

```python
# Standardize column names

df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

```python
df_clean.columns
```

```python
df_clean.head()
```

```python
# Cleaning text fields
df_clean["item"] = df_clean["item"].str.strip()
df_clean["payment_method"] = df_clean["payment_method"].str.strip()
df_clean["location"] = df_clean["location"].str.strip()
```

```python
df_clean.head()
```

```python
import numpy as np
```

```python
# Converting placeholder to missing values
invalid_values = ['UNKNOWN', 'ERROR', 'Unknown', 'unknown']

df_clean = df_clean.replace(invalid_values,np.nan)
```

```python
# Convert numeric columns
df_clean["quantity"] = pd.to_numeric(
    df_clean["quantity"],
    errors = 'coerce'
)
```

```python
df_clean["price_per_unit"] = pd.to_numeric(
    df_clean["price_per_unit"],
    errors = 'coerce'
)
```

```python
df_clean["total_spent"] = pd.to_numeric(
    df_clean["total_spent"],
    errors = 'coerce'
)
```

```python
# Convert date time
df_clean['transaction_date'] = pd.to_datetime(
    df_clean['transaction_date'],
    errors = 'coerce'
)
```

```python
df_clean['transaction_date'].isna().sum()
```

```python
# Replacing Missing Values in Categorical Columns

df_clean['item'] = df_clean['item'].fillna('Unknown') 
df_clean['payment_method'] = df_clean['payment_method'].fillna('Unknown') 
df_clean['location'] = df_clean['location'].fillna('Unknown') 
```

```python
df_clean.columns
```

```python
df_clean.head()
```

```python
# Replacing Missing Numerical Values - total_spent
# Creating temp calculated column

df_clean['calculated_total'] = (df_clean['quantity'] *
                                df_clean['price_per_unit']
)

mask = (
    df_clean['total_spent'].isna() &
    df_clean['quantity'].notna() &
    df_clean['price_per_unit'].notna()
)

df_clean.loc[mask, 'total_spent'] = (
    df_clean.loc[mask, 'quantity'] *
    df_clean.loc[mask, 'price_per_unit']
)
```

```python
# Replacing Missing Numerical Values - quantity

df_clean.loc[
    df_clean['quantity'].isna() &
    df_clean['total_spent'].notna() &
    df_clean['price_per_unit'].notna(),
    'quantity'
] = (
    df_clean['total_spent'] / df_clean['price_per_unit']
)
```

```python
# Replacing Missing Numerical Values - price_per_unit

df_clean.loc[
    df_clean['price_per_unit'].isna() &
    df_clean['total_spent'].notna() &
    df_clean['quantity'].notna(),
    'price_per_unit'
] = (
    df_clean['total_spent'] / df_clean['quantity']
)
```

```python
# Checking missing values

print("Missing Quantity:", df_clean['quantity'].isna().sum())
print("Missing Price:", df_clean['price_per_unit'].isna().sum())
print("Missing Total Spent:", df_clean['total_spent'].isna().sum())
```

```python
# Attempting to find prices based on common items

df_clean[df_clean['price_per_unit'].isna()][
    ['item', 'quantity', 'price_per_unit', 'total_spent']
]
```

```python
df_clean.groupby('item')['price_per_unit'].agg(['count', 'median', 'min', 'max'])
```

```python
# Filling missing values for price_per_unit as per median values

df_clean['price_per_unit'] = df_clean['price_per_unit'].fillna(
    df_clean.groupby('item')['price_per_unit'].transform('median')
)
```

```python
print("Missing Price:", df_clean['price_per_unit'].isna().sum())
```

```python
df_clean['quantity'] = df_clean['quantity'].fillna(
    df_clean['total_spent'] / df_clean['price_per_unit']
)
```

```python
df_clean[df_clean['quantity'].isna()][
    ['item', 'quantity', 'price_per_unit', 'total_spent']
]
```

```python
df_clean.groupby('item')['quantity'].agg(['count', 'median', 'min', 'max'])
```

```python
# Filling missing values for quantity as per median values

df_clean['quantity'] = df_clean['quantity'].fillna(
    df_clean.groupby('item')['quantity'].transform('median')
)
```

```python
# Converting quanity to integer

df_clean['quantity'] = df_clean['quantity'].round().astype(int)
```

```python
# Filling missing values for total_spent using calculations

df_clean['total_spent'] = df_clean['total_spent'].fillna(
    df_clean['quantity'] * df_clean['price_per_unit']
)
```

```python
print("Missing Quantity:", df_clean['quantity'].isna().sum())
print("Missing Price:", df_clean['price_per_unit'].isna().sum())
print("Missing Total Spent:", df_clean['total_spent'].isna().sum())
```

```python
print("Quantity <= 0:", (df_clean['quantity'] <= 0).sum())
print("Price <= 0:", (df_clean['price_per_unit'] <= 0).sum())
print("Total spent <= 0:", (df_clean['total_spent'] <= 0).sum())
```
