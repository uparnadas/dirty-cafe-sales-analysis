```python
import pandas as pd
```

```python
df_cafe = pd.read_csv("C:/DSML/Jobs/2026/Internship/SkillAudit/CafeProject/data/processed/dirty_cafe_sales_cleaned.csv")
```

```python
df_cafe.head()
```

```python
product_revenue = (
    df_cafe
    .groupby('item')
    .agg(
        total_units_sold=('quantity', 'sum'),
        total_revenue=('total_spent', 'sum')
    )
    .reset_index()
    .sort_values('total_revenue', ascending=False)
)

product_revenue['total_revenue'] = (
    product_revenue['total_revenue'].round(2)
)

print(product_revenue)
```

```python
product_units = (
    df_cafe
    .groupby('item')
    .agg(
        total_units_sold=('quantity', 'sum'),
        transaction_count=('transaction_id', 'count'),
        avg_quantity_per_transaction=('quantity', 'mean')
    )
    .reset_index()
    .sort_values('total_units_sold', ascending=False)
)

product_units['avg_quantity_per_transaction'] = (
    product_units['avg_quantity_per_transaction'].round(2)
)

print(product_units)
```

```python
location_performance = (
    df_cafe
    .groupby('location')
    .agg(
        transaction_count=('transaction_id', 'count'),
        total_units_sold=('quantity', 'sum'),
        total_revenue=('total_spent', 'sum'),
        avg_transaction_value=('total_spent', 'mean')
    )
    .reset_index()
    .sort_values('total_revenue', ascending=False)
)

location_performance['total_revenue'] = (
    location_performance['total_revenue'].round(2)
)

location_performance['avg_transaction_value'] = (
    location_performance['avg_transaction_value'].round(2)
)

print(location_performance)
```

```python
payment_performance = (
    df_cafe
    .groupby('payment_method')
    .agg(
        transaction_count=('transaction_id', 'count'),
        total_units_sold=('quantity', 'sum'),
        total_revenue=('total_spent', 'sum'),
        avg_transaction_value=('total_spent', 'mean')
    )
    .reset_index()
    .sort_values('transaction_count', ascending=False)
)

payment_performance['total_revenue'] = (
    payment_performance['total_revenue'].round(2)
)

payment_performance['avg_transaction_value'] = (
    payment_performance['avg_transaction_value'].round(2)
)

print(payment_performance)
```

```python
product_location = (
    df_cafe
    .groupby(['location', 'item'])
    .agg(
        units_sold=('quantity', 'sum'),
        revenue=('total_spent', 'sum')
    )
    .reset_index()
    .sort_values(
        ['location', 'revenue'],
        ascending=[True, False]
    )
)

product_location['revenue'] = (
    product_location['revenue'].round(2)
)

print(product_location)
```

```python
# Calculate total revenue by product
product_revenue = (
    df_cafe
    .groupby('item')
    .agg(
        total_revenue=('total_spent', 'sum')
    )
    .reset_index()
)

# Calculate the average revenue across products
average_product_revenue = product_revenue['total_revenue'].mean()

# Keep products with revenue above the average
above_average_products = (
    product_revenue[
        product_revenue['total_revenue'] > average_product_revenue
    ]
    .sort_values('total_revenue', ascending=False)
)

above_average_products['total_revenue'] = (
    above_average_products['total_revenue'].round(2)
)

print("Average revenue per product:", round(average_product_revenue, 2))
print("\nProducts with above-average revenue:")
print(above_average_products)
```

```python
# Calculate revenue by location and product
product_location_revenue = (
    df_cafe
    .groupby(['location', 'item'])
    .agg(
        revenue=('total_spent', 'sum')
    )
    .reset_index()
)

# Rank products within each location by revenue
product_location_revenue['product_rank'] = (
    product_location_revenue
    .groupby('location')['revenue']
    .rank(method='first', ascending=False)
)

# Keep the highest-revenue product from each location
top_products = (
    product_location_revenue[
        product_location_revenue['product_rank'] == 1
    ]
    .sort_values('location')
)

top_products['revenue'] = top_products['revenue'].round(2)

print(top_products)
```

```python
monthly_revenue = (
    df_cafe
    .groupby(df_cafe['transaction_date'].dt.to_period('M'))
    .agg(
        transaction_count=('transaction_id', 'count'),
        units_sold=('quantity', 'sum'),
        total_revenue=('total_spent', 'sum'),
        avg_transaction_value=('total_spent', 'mean')
    )
    .reset_index()
)

monthly_revenue['total_revenue'] = monthly_revenue['total_revenue'].round(2)
monthly_revenue['avg_transaction_value'] = monthly_revenue['avg_transaction_value'].round(2)

print(monthly_revenue)
```

```python
payment_location = (
    df_cafe
    .groupby(['location', 'payment_method'])
    .agg(
        transaction_count=('transaction_id', 'count'),
        total_revenue=('total_spent', 'sum'),
        avg_transaction_value=('total_spent', 'mean')
    )
    .reset_index()
    .sort_values(['location', 'transaction_count'], ascending=[True, False])
)

payment_location['total_revenue'] = payment_location['total_revenue'].round(2)
payment_location['avg_transaction_value'] = payment_location['avg_transaction_value'].round(2)

print(payment_location)
```

```python
product_transaction_value = (
    df_cafe
    .groupby('item')
    .agg(
        transaction_count=('transaction_id', 'count'),
        avg_product_transaction=('total_spent', 'mean')
    )
    .reset_index()
)

overall_avg_transaction = df_cafe['total_spent'].mean()

above_average_products = (
    product_transaction_value[
        product_transaction_value['avg_product_transaction'] > overall_avg_transaction
    ]
    .sort_values('avg_product_transaction', ascending=False)
)

above_average_products['avg_product_transaction'] = (
    above_average_products['avg_product_transaction'].round(2)
)

print("Overall average transaction value:", round(overall_avg_transaction, 2))
print("\nProducts with above-average transaction value:")
print(above_average_products)
```
